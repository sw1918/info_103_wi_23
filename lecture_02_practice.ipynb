The language we'll be using this quarter is called "Python."

In python, you will write lines of text called "statements", and the computer will read those statements, turn them into computer language, and execute them. For example, if you want the computer to display something, you can run the following statement:

```python
print("Hello world!")
```

The above code used the "print" function to write something out. Functions in programming are like functions in math, in that they accept something (a variable) and turn it into some output. In this case, print() is as simple as can be; we give it something, and it prints it out! 

In this case, we're giving print() a "string" (a bit of code wrapped in quotation marks like "", which represents text).

You can also do math in Python. If you want to add some numbers, subtract, multiply, etc, you can use those operators to do so:

```python
5 + 8 +2
```

Here are Python's arithmetic operators:

```python
3 + 3 # Addition
```

```python
3 - 2 # Subtraction
```

```python
3 * 3 # Multiplication
```

```python
6 / 2 # Division
```

```python
7 % 2 # Modulus
```

```python
3 ** 3 # Exponentiation
```

```python
10 // 3 # Floor division
```

```python
(10*10) - (6 / 3)
```

If you don't want a particular line of code to run, you can add the # symbol to turn everything following that symbol into a comment. Comments are useful when marking up your code, or to temporarily remove parts of your code

```python
print("This will run")
#print("This is commented, so it will not run")
print("Also, this will run")
#Let's try doing a math operation, and printing it
print(1+1+1+1)
```
