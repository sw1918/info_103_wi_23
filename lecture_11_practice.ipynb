Let's continue our Bluesky showcase and talk about feeds! A feed on Bluesky is a custom view of posts according to an algorithm developed by Bluesky developers or users. Some are quite simple ("All posts with x hashtag will display in chronological order") while some are a bit more complex ("Display a mix of popular content from your mutuals, along with content popular within your broader network").

Now, here's the trickier part. To get a feed, we need the URI of the feed, to give the get_feed() function as a parameter. I've created a function that does this for you; you just need to copy-paste the url of a feed as the parameter. 

```python
def get_did_from_url(url):
    feed_info = url.split("/profile/")[1]
    handle = feed_info.split("/feed/")[0]
    did = client.resolve_handle(handle)
    did = str(did)
    did = did[5:]
    did = did[:-1]
    return did

def get_feed_name_from_url(url):
    feed_name = url.split("/feed/")[1]
    return feed_name

def feed_uri_from_url(url):
    did = get_did_from_url(url)
    feed_name = get_feed_name_from_url(url)
    uri = f"at://{did}/app.bsky.feed.generator/{feed_name}"
    return uri
```

Let's try it out! The code below will get the first 30 posts of a given feed, and return them as a list. I've also created a bunch of helper functions to extract useful data from each post. 

Note that the objects that are being put in the list are FeedViewPost objects. These all have an attribute .post, which represents the actual post object, called a PostView. These have attributes like author, like_count, record (which includes much of the content of the post), embed (which includes things like images), etc. 

If you want to learn more, you can access definitions of all these objects here: https://atproto.blue/en/latest/atproto/atproto_client.models.app.bsky.feed.html

```python
def list_feed_posts(url):
    uri = feed_uri_from_url(url)
    data = client.app.bsky.feed.get_feed({
        'feed': uri,
        'limit': 30,
    })
    return data.feed

def get_post_text(post):
    return post.post.record.text

def get_post_author_handle(post):
    return post.post.author.handle

def get_post_like_count(post):
    return post.post.like_count

def get_post_reply_count(post):
    return post.post.reply_count

def get_post_created_at(post):
    return post.post.record.created_at

#Note that this one requires making a seperate API call
def get_post_replies(post):
    uri = post.post.uri
    thread = client.app.bsky.feed.get_post_thread({
        'uri': uri,
        #'depth': 1  # You can optionally set the depth, if you want to...
        })
    replies = []
    return thread.thread.replies
```
