Variables allow us to assign names, store values, and perform meaningful operations

```python
likes = 18
comments = 4
shares = 2

engagement = likes + (comments * 2) + (shares * 3)

print(engagement)
```

We've seen functions like print() that do helpful things. Some functions return data for us, which we can then store in new variables

```python
from math import sqrt

my_number = 9
new_number = sqrt(my_number)
print(new_number)
```
