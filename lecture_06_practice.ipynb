If statements only allow the code underneath them to run if the conditional statement returns True

```python
if 1 > 0:
	print("a")
if "a" in "apple" or 0 > 1:
	print("b")
if "a" in "apple" and 0 > 1:
	print("c")
print("d")
```

You can extend this functionality using if/elif/else statements:

```python
x = 0 # Feel free to change the value of x here

if x > 2:
    print("x is bigger than 2")
elif x > 0:
	print("Well, at least x is positive…")
else:
	print("x is not very big…")
	print("it’s either 0 or negative!")
```

While loops continue to run until their conditional statement evaluates to False.

```python
sum = 0
while sum < 10:
    sum += 5
    print(sum)
print("done")
```

Finally, you can use the break and continue keywords to control the flow of your loops. Break stops the loop entirely, while continue only skips the current iteration of the loop.

```python
sum = 0
while True:
    sum += 1
    print(sum)
    if sum > 1:
        break
    continue
    print("this will not print")
print("done")
```
