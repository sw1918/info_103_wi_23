# Practice: Looping through lists and dictionaries
## Follow along with lecture! 

You can create lists using square brackets, like this:

```python
my_list = []
```

You can create lists with items in them, like this:

```python
my_list = ["dog", "cat", "rat"]
print(my_list)
```

Access items in your list by specifying their index in square brackets. Remember, the first item in a list has an index of 0, not 1.

```python
new_var = my_list[1]
print(new_var)
```

You can add items to the end of your list using the append() function

```python
my_list.append("bat")
print(my_list)
```

You can remove items from your list using the "remove" function

```python
my_list.remove("dog")
print(my_list)
```

Note that if you have multiple of the same item in a list, only the first matching entry gets removed

```python
dogs = ["dog", "dog", "dog", "dog"]
dogs.remove("dog")
print(dogs)
```

Now, let's do some loops! To run a for loop, you need a list to iterate through, and you need to give a variable name to store each item in the last as you pass through it, like this:

```python
my_list = ["cat", "frog", "bat"]
for x in my_list:
	print("This is a " + x)
```

If you just want to do something a certain amount of times, you can use the range() function:

```python
for i in range(3):
    print("Hello!")
    print("It's nice to meet you!")
```

If you want to keep track of something while you iterate through a list (the number of items it has, for example), you can create a loop variable outside of the list. This allows you to update it as you move through the list, and access it once the list is done:

```python
total_animals = 0
my_list = ["cat", "dog", "cow"]
for a in my_list:
	total_animals += 1
print("I have " + str(total_animals) + " animals")
```
