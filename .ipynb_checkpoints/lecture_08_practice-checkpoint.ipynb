# Practice: Function definitions

## Follow along with lecture
First, you have to define a function using the "def" keyword

```python
def say_hi():
    print("Hi!")
```

Once a function is defined, you can call it like this:

```python
say_hi()
```

We can call our function wherever and whenever we want now!

```python
for i in range(5):
    say_hi()
```

We can give our functions parameters that influence their input:

```python
def repeat_hi(n):
    for i in range(n):
        print("Hi!")
```

```python
repeat_hi(3)
```

If we need multiple parameters, we can just keep adding them inside the parentheses like this:

```python
def full_greeting(first, last):
    print("Hi " + first + " " + last)
```

```python
full_greeting("Spencer", "Williams")
```

What if we want our function to produce something? We could try this...

```python
def multiply(n, m):
    n * m

print(multiply(3, 3))
```

...but, that doesn't work! By default, functions don't leave any output unless we use the "return" keyword to specify what value the function should resolve to. Let's try it:

```python
def multiply(n, m):
    product = n * m
    return product

print(multiply(3, 3))
```

Finally, be aware of scope when creating variables inside functions! If you try running the code below, you'll run into a name error telling you that "name 'y' is not defined." This is because x is in the "global scope," meaning it can be seen by other functions, whereas y is in the "local scope" of the function, meaning it can't be seen outside the function, even after the function is run. 

```python
x = 10
def my_func():
    y = 5
    return y * x
print(my_func())
print(y)
```

```python
def counter():
    for i in range(5):
        print(i)
        time.sleep(1)
```
