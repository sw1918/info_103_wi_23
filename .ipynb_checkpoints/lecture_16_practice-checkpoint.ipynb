We can allow our users to input data into our programs using python's built in input() function. It accepts a string, which will prompt the user with an instruction. The user can then type something into the input box, and press 'enter' to input:

```python
user_input = input("Type something in here:")

print("you typed in: " + user_input)
```

We can do all sorts of cool stuff with this! For example, this is basically how those old text adventure games worked. You can try this one out if you want:

```python
monster = {
    "name": "Goblin",
    "hp": 5
}

moves = [
    {
        "name": "Attack",
        "damage": 2
    },
    {
        "name": "Fire Ball",
        "damage": 5
    }
]

print("Encountered a " + monster["name"] + "!")
while monster["hp"] > 0:
    message = "Available options: "
    for m in range(len(moves)):
        message = message + moves[m]["name"]
        if m < len(moves) - 1:
            message = message + ", "
        else:
            message = message + " "
    print(message)
    user_input = input("Select an attack: ")
    for m in moves:
        found_move = False
        if user_input == m["name"]:
            print("Player 1 used " + m["name"])
            monster["hp"] = monster["hp"] - m["damage"]
            print("It did " + str(m["damage"]) + " damage!")
            found_move = True
            break
    if found_move == False:
        print("Didn't recognize that move; please try again!")

print("Defeated the goblin!")
```
